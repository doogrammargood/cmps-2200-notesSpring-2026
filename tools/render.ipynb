```python
from latexrender import LatexRender
# Flow Syntax
L=[1,2,3,4,5]
render = (
    LatexRender()
    .render(
        L,
        color='cyan',
        bold=True,
        underline=True,
        italic=[False, False, True, True, False],
    )
    .render('num=1')
    .render(
        [9,2,3,4,5],
        color=[None, 'green', 'red', None, 'blue'],
        bold=[True, False, False, False, True],
        underline=[True, False, False, False, True],
        deleteline=[False, False, True, False, False],
        italic=[False, False, True, True, False],
    )
    .render('end')
)
# Save to file
render.to_latex('out.tex')
render
```

```python
# Modulized Syntax
render = LatexRender()
render.render([1,2,3,4,5], color='cyan', bold=True, underline=True)
render
```

## This week

We will develop a heirarchy of graph classes. Interval graphs $\subset$ chordal graphs $\subset$ perfect graphs.

We discuss algorithms for recognizing each class of graph.

Our recognition algorithm for perfect graphs will lead to a structural decomposition of those graphs. By taking advantage of this structural decomposition, we get several fast algorithms.

By adding edges to arbitrary graphs to make them chordal, we can find good algorithms to calculate many graph properties. This is a general strategy for constructing fast algorithms.

## Final goal

We will analyze this general strategy. We will make the point that the independence number cannot be calculated efficiently under our current knowledge of algorithms. Therefore, the boundary between the general strategy being efficient and inefficient reveals a fundamental barrier to computation.

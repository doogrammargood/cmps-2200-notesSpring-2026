# Graph Structures

Here, we review some of the concepts and terminology related to sub-structures of graphs. Throughout, let $G$ be a simple graph.

## Subgraphs

A subgraph of $G$ is a graph $G^\prime$ such that $V(G^\prime)\subseteq V(G)$ and $E(G^\prime)\subseteq E(G)$.

## Induced Subgraphs

An induced subgraph of $G$ is a type of subgraph $G^\prime$ that has the additional property that if $e\in E(G)$ and $v_1,v_2 \in V(G^\prime)$ are the endpoints of $G$ then $e\in E(G^\prime)$. In other words, induced subgraphs of $G$ are determined by a subset of $V(G^\prime)\subseteq V(G)$, and the resulting edges are all edges of $G$ whose endpoints are both in $V(G^\prime)$.

## Topological Minors

A topological minor of $G$ is a graph $G^\prime$ that can be formed by taking a subgraph of $G$, and contracting edges that are incident to vertices of degree $2$.

## Minors

A minor of $G$ is a graph $G^\prime$ that can be formed by deleting edges, vertices, and contracting edges.

# Graph Types

In this section, we describe three classes of graphs. Each class is a subclass of the next class. These types of graphs are important because they are general classes of graphs for which we have good algorithms to calculate the independence number.

## Interval Graphs

A graph $G$ is an *interval graph* if its vertices can be identified with intervals $v_i = (s_i, f_i)\subset \mathbb{R}$ such that vertices are adjacent iff the corresponding intervals intersect.

## Chordal Graphs

A graph $G$ is *chordal* if it does not have any induced cycles of length $4$ or greater.

Alternate characterization: A graph $G$ is chordal iff the vertices can be ordered $V(G)=v_1,v_2,\dots,v_n$ such that for all vertices $v_i$, $\mathcal{N}(v_i) \cap \{v_1,v_2,\dots,v_{i-1}\}$ induces a clique in $G$. Such an ordering of the vertices is called a *perfect elimination ordering*.

## Perfect Graphs

A graph $G$ is *perfect* if neither $G$ nor $\overline{G}$ has any induced odd cycles.

Alternate characterization: $G$ is perfect iff for every induced subgraph $G^\prime$ of $G$, $\omega(G)=\chi(G)$, where $\omega$ and $\chi$ are the clique number and chromatic numbers, respectively.

## Recognizing Chordal Graphs

[source](https://www.youtube.com/watch?v=txaGsawljjA)

Knuth describes a beautiful algorithm to recognize when a graph is chordal. The algorithm demonstrates that chordal graphs can be alternatively characterized as having a perfect elimination ordering.

The algorithm is a special type of "greedy" search of the graph, which is commonly called "maximum cardinality search." 

The search algorithm starts at an arbitrary source vertex, then repeatedly visits the next vertex with the largest amount of visited neighbors, breaking ties arbitrarily. If the graph is chordal, then the resulting order of the vertices is a perfect elimination ordering. If the graph is not chordal, then we will notice when we visit a vertex whose neighbors do not form a clique.

## Structure of chordal graphs

The greedy search algorithm and resulting perfect elimination order have deep consequences for the structure of chordal graphs. 

At each step of the algorithm, we can keep track of the maximal cliques. When we add another vertex $v$ to the perfect elimination ordering, we update the maximal cliques by first finding the neighbors of $v$ that occur before $v$ in the ordering. The those previous neighbors form a clique. If it is a maximal clique, then we add $v$ to it. Otherwise, we create a new maximal clique that contains $v$ and its neighbors.

At each step, we either create a new maximal clique or enlarge an existing one. This demonstrates that the number of maximal cliques in a chordal graph $G$ is at most $|V(G)|$.

## Tree structure of maximal cliques

As we progress through the greedy search algorithm, we can also construct a tree structure to describe how the maximal cliques are attached. We maintain a tree $T$ of maximal cliques that is initially empty. Each time we add a new maximal clique, we add a vertex to $T$ and an edge from the new vertex to an existing vertex that contains all of the old vertices. ## Chordal Completion

Given a connected simple graph $G$, a *chordal completion* of $G$ is a chordal graph on the same vertex set as $G$ that has $G$ as a subgraph. Here, we show how the greedy search algorithm can be used to calculate a chordal completion. The point is that we will use the clique tree of the chordal completion to structure computations about our original graph $G$.

The algorithm begins by ordering the vertices of $G$ using the greedy search algorithm. Then, we add additional edges such that the resulting graph passes our chordal recognition test from the previous section.

There is a catch with this: we need to be careful to add edges in such a way that the new edges do not change the ordering of the vertices that results from the greedy search. The trick is to loop through the vertices in reverse. For each vertex, add an edge between the all pairs of previous neighbors of that vertex. Notice that if we loop through the vertices in the order that they were visited by the greedy search, then the new edges that we add may change the order that the vertices should be visited by a greedy search. 

We also need to be careful to include the newly-added edges from previous vertices when determining the previous neighbors of a vertex.

## Dynamic Programming

We can write out the steps of the dynamic program for the example.

<img src="./figures/full_example_screenshots/Screenshot from 2026-04-28 17-54-51.png" width="100%"/>

Our table is indexed by vertices of the tree (```j```) and independent sets $S$ of the graph restricted to the bag of ```j```. The value of ```c[j][S]``` is the maximal size of an independent set among the subgraph restricted to the unions of the bags from ```j``` away from the root whose restriction to the bag of ```j``` is exactly ```S```.

For each tree node, we will write the set of independent sets on the graph restricted to that bag as $\mathcal{I}_k$. Let us precompute these values.

\begin{align*}
\mathcal{I}_1 &= \{\emptyset,\{1\}, \{4\},\{5\}, \{1,4\}\}\\
\mathcal{I}_3 &= \{\emptyset, \{1\}, \{2\},\{3\}, \{1,3\}\}\\
\mathcal{I}_2 &= \{\emptyset, \{1\}, \{3\},\{4\}, \{1,3\}, \{1,4\}\}\\
\end{align*}

Thus, our dynamic programming table will have $16$ entries.

Note that we have listed the bags in a topological order, meaning that when the tree node $X2$ is the root, children are listed before their parents.


\begin{array}{c|cccccc}
\hline
\mathcal{I}_1 & \emptyset & \{1\} & \{4\} & \{5\} & \{1,4\} \\
              & - & - & - & - & - \\
\hline
\mathcal{I}_3 & \emptyset & \{1\} & \{2\} & \{3\} & \{1,3\}\\
              & -& - & - & - & - \\
\hline
\mathcal{I}_2 & \emptyset & \{1\} & \{3\} & \{4\} & \{1,3\} & \{1,4\} \\
              & - & - & - & - & - & - \\
\end{array}


The first two rows of the table can be filled in easily.
\begin{array}{c|cccccc}
\hline
\mathcal{I}_1 & \emptyset & \{1\} & \{4\} & \{5\} & \{1,4\} \\
              & 0 & 1 & 1 & 1 & 2 \\
\hline
\mathcal{I}_3 & \emptyset & \{1\} & \{2\} & \{3\} & \{1,3\}\\
              & 0& 1 & 1 & 1 & 2 \\
\hline
\mathcal{I}_2 & \emptyset & \{1\} & \{3\} & \{4\} & \{1,3\} & \{1,4\} \\
              & - & - & - & - & - & - \\
\end{array}

When we calculate the values for $\mathcal{I}_2$, we need to refer to the previously computed values to account for how the largest independent intersects with the bags $\mathcal{I}_1$ and $\mathcal{I}_3$.

For example, to calculate ```c[2][{}]```, we must find the size of the largest independent $X$ set that does not intersect the bag $\{1,3,4\}$. We consider all possibilities for how $X$ can intersect bag $\{1,4,5\}$, given that $X$ does not contain $1$,$3$ or $4$. These possibilities are $\emptyset$ and $\{5\}$. We refer to ```c[1][{}]``` and ```c[1][\{5\}]``` to find the size of the maximal size of $Y$ intersected with the bag $\{1,4,5\}$ and its children. We find that ```c[1][\{5\}]=1``` is the largest possibility.

We also must consider how large $X$ can intersect the bag $\{1,2,3\}$ so that we can refer to the table $c$ to find the size of $X$ intersected with $\{1,2,3\}$ and its children. Since $X$ does not intersect $\{1,3,4\}$, the only possibilities for $X$'s intersection with $\{1,2,3\}$ are $\emptyset$ and $\{2\}$. We refer to ```c[3][{}]``` and ```c[3][\{2\}]``` and find that ```c[3][\{2\}]=1``` is larger. Therefore, we can conclude that ```c[2][{}]=2```.

Let us calculate another value of the table to illustrate how overlap of the bags affects calculations.

We calculate ```c[2][{1,3}]```. We must find the size of the largest independent set $Y$ whose intersection with the bag $\{1,3,4\}$ is exactly $\{1,3\}$. Again, we consider all possibilities for how $Y$ can intersect the bag $\{1,4,5\}$ and its children. The possibilities for this intersection are $\{1\}$. We reference the table for entry ```c[1][{1}]=1```. This is the largest possible size of $Y$ intersected with the children of $\{1,4,5\}$.

We must consider the possibilities for how $Y$ can intersect the bag $\{1,2,3\}$ and reference the table to obtain the largest possible intersection of $Y$ with the bag $\{1,2,3\}$ and its chlidren. The only possibility is $\{1,3\}$. We reference the table for entry ```c[3][{1,3}]=2```.

To fill in the value of ```c[2][{1,3}]```, we apply the inclusion/exclusion principle. This gives us $2+1+2-1-2+0=2$. Therefore, we set ```c[2][{1,3}] = 2```.

In a similar way, we can fill in the rest of the table.

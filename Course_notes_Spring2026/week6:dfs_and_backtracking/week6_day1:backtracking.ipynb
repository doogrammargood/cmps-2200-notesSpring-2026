# Backtracking

## DFS

Depth first search is a special case of a more general algorithmic idea, backtracking. Or, viewed through another lens backtracking is depth-first search applied to an abstract search graph.

## Backtracking General Description.

Suppose that we have a problem where we need to make a series of choices, $c_0,c_1,\dots,c_n$. These choices need to satisfy some criterion (goal) at the end. For example, if the choices can be the neighbor to visit next in a path through a graph, and our goal may be to end at a particular vertex.

We also assume that we can check whether the choices we have made up to a point $c_0,c_1,\dots,c_k$ can possibly lead to the goal. For example, we can check whether all neighbors have been visited. If so, then then we backtrack and try a new choice for $c_{k-1}$. This may cause us to backtrack again.

## Sudoku

Sudoku is a common example for the backtracking techinque. 

[rules source](https://sudoku.com/sudoku-rules/)

- Sudoku grid consists of 9x9 spaces.
- You can use only numbers from 1 to 9.
- Each 3×3 block can only contain numbers from 1 to 9.
- Each vertical column can only contain numbers from 1 to 9.
- Each horizontal row can only contain numbers from 1 to 9.
- Each number in the 3×3 block, vertical column or horizontal row can be used only once.
- The game is over when the whole Sudoku grid is correctly filled with numbers.

### Sudoku as a backtracking problem.

We number the cells from left to right, top to bottom. We progress through the cells and try to fill in values, backtracking when we get stuck. This is an example of a backtracking problem, where our choices are the possible values to fill in for the next square. The goal is to fill in all of the squares. For any partial solution, we can check whether the partial solution obeys the Sudoku rules. If it does not, then we must backtrack. If it does, then we continue filling in the next cells.

 When we get stuck, we erase the current cell and backtrack to the previous cell. We try the next number in the previous cell. If this fails, we may need to backtrack again. In this case, we need to backtrack twice, because the previous cell was already $9$, the largest digit. We increment the current cell from $3$ to the next number that obeys the Sudoku rules, in this case $4$.


<img src="./figures/Screenshot from 2026-03-15 20-21-44.png" width="25%"/>

Then we continue filling in the next cells with the first value that obeys the Sudoku constraints.

Now we are stuck again. Since $9$ is the only number that obeys the row constraints, but it does not obey the column constraints, there is no option for this box that satisfies the Sudoku constraints. We backtrack again.

This time, we only need to backtrack one cell, because $3$ can be increased to $9$, and all of the Sudoku rules for that cell are satisfied.

```python
import math
import random

from IPython.display import display, HTML
#This code, generated by LLM, will help print each step.
def display_array(arr, red = [], blue = [], bolded = []):
    '''Input: arr is the array that we are displaying. 
            red is a list of indices to make red.
            blue is a list of indicies to make blue.
            bolded is a list of indices to bold.
        Output: None.
        Side-Effects: displays arr, with the stylings indicated.
    '''
    cells = []
    for index, x in enumerate(arr):
        if index in red and index in bolded:
            cells.append(
                f"<span style='font-weight:1000; color:darkred'>{x}</span>"
            )
        elif index in red:
            cells.append(
                f"<span style='font-weight:bold; color:darkred'>{x}</span>"
            )
        elif index in bolded:
            cells.append(
                f"<span style='font-weight:1000'>{x}</span>"
            )
        else:
            cells.append(str(x))
    html = (
        "<span style='font-family: monospace'>[ "
        + ", ".join(cells)
        + " ]</span>"
    )
    display(HTML(html))
```

<h3> SelectionSort</h3>

Here, we introduce the selection sort algorithm.

<h5>Strategy</h5>

In Selectionsort, we "select" the smallest element, then swap that element to the front of the list. Then, we select second-smallest element and swap it to its correct place. Continuing in this way, we put each element of the list in its proper place.

<h5> Implementation</h5>

Here is an implementation of selectionsort.

```python
def selection_sort(arr: list[int], quiet_mode: bool = True) -> None:
    '''
    Input: arr (a list of integers).
           quiet_mode (a boolean, whether or not to print intermediate steps)
    Output: None
    Side-Effects: mutates arr so that it is sorted.
    Repeatedly finds the smallest incorrectly-placed element and swaps it to the correct position.
    '''
    n= len(arr)
    for pass_number in range(n-1):
        if not quiet_mode:
            print(f"{pass_number=}")
        min_candidate = arr[pass_number]
        min_candidate_index = pass_number
        for i, candidate in enumerate(arr[pass_number+1:], start = pass_number+1): #This loop finds the smallest element after the first pass_number number of candidates.
            if not quiet_mode:
                display_array(arr, red = [i, min_candidate_index], bolded = list(range(0,pass_number)))
            if min_candidate > candidate:
                min_candidate_index, min_candidate = i, candidate
        arr[pass_number], arr[min_candidate_index] = arr[min_candidate_index], arr[pass_number]
    if not quiet_mode:
        display_array(arr, bolded = list(range(0,n-1)))
#Check that the sorting works with an example.
list_to_sort = [9,8,12,9,16,5,3,2]
print("before sorting", f"{list_to_sort=}")
selection_sort(list_to_sort)
print("after sorting", f"{list_to_sort=}")
assert list_to_sort == sorted(list_to_sort)
```

<h3>Correctness</h3>

After each pass of the algorithm, the number of correctly-placed elements at the beginning of the list is the number of passes. This observation is called a <i>loop variant</i>. If we wanted to dig deeper, we could prove this observation by induction. If $n$ is the length of the initial list, then after $n-1$ passes, the first $n-1$ elements will be in their correct places. This will force the last element to be in its correct place.

<h3>Insertion Sort</h3>

We introduce the insertion sort algorithm.

<h5>Strategy</h5>

We keep the first several elements of the list sorted with respect to each other. We then repeatedly "insert" new elements into the partially sorted list. After each element is sorted with respect to each other element, the entire list is sorted.

<h5>Implementation</h5>

Here is an implementation of insertion sort.

```python
def insertion_sort(arr: list[int], quiet_mode: bool=True) -> None:
    '''
    Input: arr (a list of integers).
       quiet_mode (a boolean, whether or not to print intermediate steps)
    Output: None
    Side-Effects: mutates arr so that it is sorted.
    Repeatedly inserts the next element of arr into the sorted portion of the list.
    '''
    n = len(arr)
    for pass_number in range(n):
        current_index = pass_number
        if not quiet_mode:
            print(f"{pass_number=}")
            display_array(arr, red=[current_index, current_index-1], bolded = list(range(0,pass_number)) )
        while current_index >0 and arr[current_index] < arr[current_index-1]:
            arr[current_index], arr[current_index-1] = arr[current_index-1], arr[current_index]
            current_index-=1
            if not quiet_mode:
                 display_array(arr, red = [current_index, current_index-1], bolded = list(range(0,pass_number)) )
#Check that the sorting works with an example.
list_to_sort = [9,8,12,9,16,5,3,2]
print("before sorting", f"{list_to_sort=}")
insertion_sort(list_to_sort)
print("after sorting", f"{list_to_sort=}")
assert list_to_sort == sorted(list_to_sort)
```

<h3>Summary of loop variants</h3>

Here is a table of the three basic sorting algorithms that we have seen, together with the loop variants that justify their correctness

| Algorithm | Description | Loop Variant after $k$ passes |
|-----------|-------------|--------------|
|Bubblesort |Repeatedly swap adjacent, out of order elements.| The last $k$ elements are in the correct place.|
|Selectionsort| Repeatedly find the smallest unplaced element and swaps it to its correct place. | The first $k$ elements are in their correct place.|
|Insertionsort|Repeatedly move the next element of the list into a sorted portion of the list. | The first $k$ elements are sorted relative to each other.

<h3>Recursive Formulations</h3>

We will see that algorithms are easier to analyze when the are expressed with recursion. Each of the three basic sorting algorithms can be easily expressed with recursion, as we now explain.

```python
def recursive_bubblesort(arr: list[int], pass_number: int) -> None:
    '''
    Input: arr, a list of integers. pass_number, an integer between 0 and n-1.
    Output: None
    Side-Effects: sorts arr.
    Note: A recursive implementation of the bubblesort strategy.
    '''
    if pass_number == len(arr)-1: #in this case, the list is already sorted.
        return
    else:
        for j in range(0,len(arr)-pass_number-1):
            if arr[j]>arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
        recursive_bubblesort(arr, pass_number+1)

def recursive_selectionsort(arr: list[int], pass_number: int) -> None:
    '''
    Input: arr, a list of integers. pass_number, an integer between 0 and n-1.
    Output: None
    Side-Effects: sorts arr.
    Note: A recursive implementation of the selectionsort strategy.
    '''
    if pass_number == len(arr)-1: #in this case, we have already sorted n-1 elements, so the last element must be in its place and the list is already sorted.
        return
    else:
        min_candidate = arr[pass_number]
        min_candidate_index = pass_number
        for i, candidate in enumerate(arr[pass_number+1:], start = pass_number+1): #This loop finds the smallest element after the first pass_number number of candidates.
            if min_candidate > candidate:
                min_candidate_index, min_candidate = i, candidate
                arr[pass_number], arr[min_candidate_index] = arr[min_candidate_index], arr[pass_number]
        recursive_selectionsort(arr, pass_number +1)

def recursive_insertion_sort(arr: list[int], pass_number: int) -> None:
    '''
    Input: arr, a list of integers. pass_number, an integer between 0 and n-1.
    Output: None
    Side-Effects: sorts arr.
    Note: A recursive implementation of the insertionsort strategy.
    '''
    if pass_number == len(arr): #in this case, the entire list is already sorted.
        return
    else:
        current_index = pass_number
        while current_index >0 and arr[current_index] < arr[current_index-1]:
            arr[current_index], arr[current_index-1] = arr[current_index-1], arr[current_index]
            current_index-=1
        recursive_insertion_sort(arr, pass_number+1)
#--basic check of correctness.
recursive_methods = [recursive_bubblesort, recursive_selectionsort, recursive_insertion_sort]
for sorting_method in recursive_methods:
    list_to_sort = [9,8,12,9,16,5,3,2]
    sorting_method(list_to_sort,0)
    assert list_to_sort==sorted(list_to_sort)
print("passed")
```

<h3>Recursion for analysis. Iteration for Python.</h3>

Recursive functions are not recommended in Python, because each recursive call requires setting up a function to add to the call stack. This is very slow. In fact, Python has a recursion limit (usually set to about 3000) that bounds the recursion depth. The recursive methods that we defined will fail on a list of length larger than the recursion depth limit, as we now demonstrate.

```python
arr = [random.randint(0, 99) for _ in range(3_000)] #creates a random array of length 3_000, beyond the recursion-depth limit.
print(f"Before sorting, {arr=}")
recursive_bubblesort(arr,0)
print(f"After sorting, {arr=}")
```

<h3>Analysis of recursive algorithms</h3>

Recursion allows us to analyze algorithms, even if it is not the recommended implementation in Python. The recursive formulation of the algorithm provides us with a recursive equation for the number of operations involved.

<h3>Recursive analysis of bubblesort</h3>

Let's use the recursive formulation of bubblesort to count the number of comparisons and swaps used. We should recover the formula that we found in last week's lecture on bubblesort. Let $B(n)$ be the number of comparisons used to sort a list of length $n$ using bubblesort. We find that $B(0)=0$ and that $B(n)=(n-1)+B(n-1)$. Unwinding the recursion gives us that $B(n)=(n-1)+(n-2)+(n-3)+\dots+1+0= \sum_{i=0}^{n-1}n-1-i = \frac{(n-1)n}{2}$. This exactly agrees with our previous analysis. Moreover, the recursive formulation allows us to prove the correctness of the formula by induction.

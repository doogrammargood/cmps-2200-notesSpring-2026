# Data Structures

A data structure is a structure for holding data. Since algorithms transform input data into outdata, we should review some basic data structures.

We give a few examples of common data structures and some of the allowed operations on them.

## Arrays

[source](https://www.geeksforgeeks.org/dsa/array-data-structure-guide/)

Arrays are data structures that are familiar from our study of sorting. They should be thought of as RAM (random access memory) or as a C-array. They are a collection of memory cells, each of which can be accessed in constant time.

### Example

Here is an example of an array:
| 1 | 5 | 8 | 'a' | 16 |
|---|---|---|-----|----|
| 0 | 1 | 2 | 3   | 4  |

### Operations

| Operation | Meaning | Example | Cost | Notes |
|-----------|---------|---------| -----|-------|
| Index     | Returns the element at a given index. | ```Index(2)``` = 8 | O(1)|  |
| Edit      | Changes the value in the array | ```Edit(2,8)``` changes the index-2 value to 8. | O(1) | |
| Insert    | Adds a new value into the array | ```Insert(2,'x')``` would insert 'x' to index 2. The 8 currently at index 2 would move to index 3, and all subsequent values would have their index incremented. | O(n) |  |
| Delete | Removes a value from the array. | ```Delete(2)``` would remove 8 from the array. All subsequent values would have their index decremented. | O(n) | It is more efficient to edit the value to be a special null value |
| Swap | Swaps the indices of two values in the array. | ```Swap (2,3)``` would swap the 8 and the 'a'. | O(1) | |
| Compare| Compares the values at two indices in the array | ```Compare(1,2)``` would indicate that 5<8| O(1)| We use two calls to Index. |
| Search | Finds the index of a value | ```Search('a')``` returns 3. | O(n) | If the array is unsorted, we must examine all n values. If the array is sorted, binary search takes O(log(n)) time.|

### Strengths and Weaknesses

#### Strengths:

Arrays are simple and fast. They are directly implemented by RAM.

#### Weaknesses:
Arrays are rigid. It is costly to edit them.

Memory for an array must be allocated upfront. You are not allowed to append to an array. Instead, you need to instantiate a new array.

## Linked Lists

[source](https://www.geeksforgeeks.org/dsa/linked-list-data-structure/)

Linked lists consist of a collection of nodes, and a special node known as the head. Each node contains a pointer to the next node. The addresses are memory addresses that are assigned by the operating system.

## Example

Here is an example of a linked list.

|address | value | next address|
|--------|-------|-------------|
| 0      | 1     | 678         |
---

|address | value | next address|
|--------|-------|-------------|
| 678    | 5     | 529         |
---

|address | value | next address|
|--------|-------|-------------|
| 529    | 8     | 901         |
---

|address | value | next address|
|--------|-------|-------------|
| 901    | 'a'   | 117         |
---

|address | value | next address|
|--------|-------|-------------|
| 117    | 16    | None        |
---

We can also represent this linked list using a directed graph.

### Operations

| Operation | Meaning | Example | Cost | Notes |
|-----------|---------|---------| -----|-------|
| Edit      | Change the value of a given node | ```Edit(529,'x')``` would change the 8 to an 'x'| O(1) | The hard part is finding the relevant node to edit.|
| Insert    | Insert a new node after a given node. | ```Insert(529,'x')``` would create a new node with an arbitrary address (say 662). Node 529's next address would change to 662, and 662's next address would be set to 901 | O(1) | The hard part is finding where to insert.|
| Delete | Deletes the node after a given node. | ```Delete(529)``` would delete the node 901, and would the next address of 529 to 117.| O(1) | If we wanted to delete 117 by calling 117, then we would need to search for the previous node, 901, to update its next address. This would be slow.|
| Swap | Swaps the values of two given nodes | ```Swap(678,901)``` would swap the 5 and the 'a'.| O(1) | This just takes 2 calls to Edit.|
|Compare| Compares the values of two given nodes | ```Compare(0,529)``` would indicate that 1<8| O(1)| The work of comparing is really outside of the data structure.|
|Search | Finds the address of a node containing a given value | ``` Search('a')``` would return 901| O(n) | Even if the list is sorted, we may be searching for the last element or an element that is not present. This forces us to traverse the entire linked list. |

### Variants

Linked lists can be doubly linked, meaning that each node also stores the address of its predecessor. We can also keep references to any particular nodes. For example, if we expect to reference the last element of the linked list (the tail) often, we may include a pointer to it. These modifications will not affect the asymptotic cost of the operations.

### Strengths and Weaknesses


#### Strengths

Linked lists are more flexibile than arrays. They can be edited easily. It is easier to move references to data than data itself.

#### Weaknesses

Linked lists must be traversed, and it takes time to follow each link. In addition, pointers can become confusing and can cause bugs.

## Python Lists

### Example

```my_list = [8,9,3,'x',0]```

[source](https://wiki.python.org/moin/TimeComplexity)

Python lists (as implemented in CPython) should be viewed as an array of pointers. The arrays have fixed size. Whenever an element is appended to a list whose underlying array is full, a new array that is twice as large is allocated.

### Operations

| Operation | Meaning | Example | Cost | Notes |
|-----------|---------|---------|------|-------|
| get (Index)| returns the element at a given index | ```my_list[2]``` is ```3```| O(1)| |
| set (Edit)| changes the value in the list at a given index | After ```my_list[2]=4```, the value of ```my_list``` is ```[8,9,4,'x',0]```| O(1) | |
| Insert | inserts a value into a list at a given index | After ```my_list.insert(2,5)``` the value of ```my_list``` changes to ```[8,9,5,4,'x',0]```| O(n)| As with arrays, we need to move lots of elements. |
| append | Appends a new value to the end of the list | After ```my_list.append(7)```, the value of ```my_list``` changes to ```[8,9,3,'x',0,7]```| O(1) | Fast appending is a key feature of Python lists.|
| delete | Deletes an element of the list | After ```del my_list[2]```, the value of ```my_list``` changes to ```[8,9,'x',0]```| O(n)| We need to move elements after deleting the element.|

## Skip Lists

Skip lists are a randomized data structure. They combine the ease of access of arrays with the flexibility of linked lists.

[source (recommended video)](https://www.youtube.com/watch?v=NDGpsfwAaqo)

Skip lists are linked lists with additional long-range links that allow you to traverse the list quickly. The long-range links are arranged at random according the process described in the video. The video also provides analysis for the costs of the operations in this table. These are expected costs, averaged over the randomness in the creation of the skip list.

### Operations

| Operation | Meaning  | Expected Cost | Notes |
|-----------|----------|---------------|-------|
|search (when sorted)| Assuming the contents of the skip list are sorted, find the address of a value. | O(log(n)) | |
|Insert | Insert a new element just after a given node of the skip list. | O(log(n)) | This is harder than inserting into a linked list, because the long-range links must be managed.|
|Delete | Delete a given node from the skip list. | O(log(n))|

## Hash Tables

[source](https://www.geeksforgeeks.org/dsa/hash-table-data-structure/)

A hash table is a mapping from keys to values, so that we can quickly retrieve the value associated with a given key. This is achieved by using a <i>hash function</i> to compute an address from the key, and storing the value in the computed address.

A good hash function is a function from the set of possible keys to the set of possible addresses. Typically, there are many more keys than addresses. This creates the potential for a <i>hash collision</i>, where two keys are mapped to the same address by the hash function. A hash table must have a protocol to handle this scenario. When there are many more addresses than used keys, the probability of a collision is small. This fact is called the <i>birthday paradox</i>.

Hash tables are an ongoing area of research in Data Structures.


### Operations

| Operation | Meaning  | Expected Cost | Notes |
|-----------|----------|---------------|-------|
| get       | Retrieve the value associated with a key| O(1) | |
| set       | Add a key/value pair to the hash table | O(1) | |

### Strengths and weaknesses

#### Strengths:
Hash tables can scale quickly. Some popular databases like Redis are essentially just big hash tables.

#### Weaknesses:
If users can store information in hash tables, then they can intentionally engineer collisions and harm performance. The get function involves computing a hash function, and then accessing an address. It will not be as fast as directly accessing the address, as in an array.

## Stacks

A stack is a data structure that supports two operations, push (add to the top) and pop (remove from the top).

Stacks are used to keep track of currently active set of function calls in Python. When a function is called, it is added to the stack. When a function returns, it is popped off of the stack. You can view the stack using Python's ```inspect``` module.

[source](https://www.geeksforgeeks.org/dsa/stack-data-structure/)

### Operations

| Operation | Meaning  | Expected Cost | Notes |
|-----------|----------|---------------|-------|
| push       | add a new element to the top of the stack| O(1) | |
| pop       | remove an element from the top of the stack | O(1) | |

### Strengths and Weaknesses

#### Strengths

Stacks are simple. Because only the top element needs to be accessed, we can store the lower elements in external memory.

#### Weaknesses

It is difficult to access an item in the middle of the stack.

## Queues

A queue is a data structure that supports two operations, enqueue (add to the back) and dequeue, (remove from the front).

### Operations

| Operation | Meaning  | Expected Cost | Notes |
|-----------|----------|---------------|-------|
| enqueue       | add a new element back of the queue| O(1) | |
| dequeue       | remove an element from the front of the queue | O(1) | |

### Strengths and Weaknsesses

#### Strengths
Queues are simple and intuitive. Because elements in the middle of the queue are not accessed, the middle of the queue can be stored in external memory.

#### Weaknesses
It is difficult to access an element in the middle of a queue.
